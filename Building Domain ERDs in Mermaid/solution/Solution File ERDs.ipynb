# Building Domain ERDs in Mermaid

## Background

Below you will find a data contract for a 4-table database that tracks B2B (Business-to-Business) sales of kitchen equipment for the fictional company **ABC Kitchen Corp**.

The tables are:

- **Inventory**
- **Salesperson**
- **Customer**
- **Purchases**

All information, including column names, data types, validation rules, and key relationships, is defined in the data contract below.

---

## Assignment

Using the information found in the above data contract, create **three ERDs using Mermaid**.

---



### Conceptual ERD

The Conceptual ERD should:

- Include only entity names
- Show relationship lines
- Exclude attributes
- Exclude data types
- Exclude primary and foreign keys

---

```mermaid
erDiagram

    INVENTORY
    CUSTOMER
    SALESPERSON
    PURCHASES

    SALESPERSON ||--o{ CUSTOMER : manages
    CUSTOMER ||--o{ PURCHASES : makes
    INVENTORY ||--o{ PURCHASES : contains
```

### Logical ERD

The Logical ERD should:

- Include entities and attributes
- Use non-technical terminology
- Use generic data types such as:
  - Text
  - String
  - Number
  - Decimal
  - Date

---


```mermaid
erDiagram

    INVENTORY {
        String Inventory_ID
        String Product_Name
        String Category
        Decimal Unit_Price
        Integer Quantity_On_Hand
    }

    SALESPERSON {
        String Salesperson_ID
        String First_Name
        String Last_Name
        String Email
        String Territory
    }

    CUSTOMER {
        String Customer_ID
        String Company_Name
        String Industry
        String State
        String Salesperson_ID
    }

    PURCHASES {
        String Purchase_ID
        Date Purchase_Date
        String Customer_ID
        String Inventory_ID
        Integer Quantity_Purchased
        Decimal Total_Amount
    }

    SALESPERSON ||--o{ CUSTOMER : assigned_to
    CUSTOMER ||--o{ PURCHASES : places
    INVENTORY ||--o{ PURCHASES : purchased_in
```

### Physical ERD

The Physical ERD should:

- Be designed for PostgreSQL
- Use computer-friendly naming conventions
- Use PostgreSQL-compatible data types
- Include primary keys and foreign keys
- Show all relationships

Examples of PostgreSQL data types:

- VARCHAR
- INTEGER
- NUMERIC
- DATE
- CHAR

---

```mermaid
erDiagram
    INVENTORY {
        VARCHAR(20) inventory_id PK
        VARCHAR(150) product_name
        VARCHAR(75) category
        NUMERIC(10_2) unit_price
        INTEGER quantity_on_hand
    }
    SALESPERSON {
        VARCHAR(20) salesperson_id PK
        VARCHAR(75) first_name
        VARCHAR(75) last_name
        VARCHAR(150) email
        VARCHAR(100) territory
    }
    CUSTOMER {
        VARCHAR(20) customer_id PK
        VARCHAR(150) company_name
        VARCHAR(75) industry
        CHAR(2) state
        VARCHAR(20) salesperson_id FK
    }
    PURCHASES {
        VARCHAR(20) purchase_id PK
        DATE purchase_date
        VARCHAR(20) customer_id FK
        VARCHAR(20) inventory_id FK
        INTEGER quantity_purchased
        NUMERIC(12_2) total_amount
    }
    SALESPERSON ||--o{ CUSTOMER : assigned_to
    CUSTOMER ||--o{ PURCHASES : places
    INVENTORY ||--o{ PURCHASES : purchased_in
```
