# Exercise: Match the Data Product to the Right AWS Database

## Scenario

You are a **data architect** at a SaaS company. Several product and business teams have created small data products and provided sample data.

Your job is to review each data product scenario, inspect the sample data format, evaluate the access patterns, and decide which AWS database or data service is the best fit.

---


## Learning Objective

By the end of this exercise, you should be able to distinguish between AWS database and data services based on:

- Data structure
- File format
- Access pattern
- Query behavior
- Transactional versus analytical workload
- Need for semantic search or retrieval-augmented generation



## Available AWS Database Options

| AWS Service | Best For |
|---|---|
| Amazon RDS PostgreSQL | Structured relational data with SQL, joins, primary keys, foreign keys, constraints, and transactional workloads |
| Amazon DynamoDB | High-volume key-value or document-style access patterns with predictable lookups and low-latency reads/writes |
| Amazon Redshift | Analytical workloads, aggregations, dashboards, reporting, and large-scale OLAP queries |
| Amazon OpenSearch Serverless / Amazon Bedrock Knowledge Bases | Semantic search, Retrieval-Augmented Generation (RAG), embeddings, document retrieval, and natural language question answering over unstructured text |




For each provided sample dataset:

1. Inspect the file format.
2. Read the scenario.
3. Identify the primary access pattern.
4. Compare at least two possible AWS service choices.
5. Select the best AWS database or data service.
6. Explain why the selected service is the best fit.

# Data Product B: User Activity Events

## Scenario

The Product Analytics team tracks application events such as page views, button clicks, login events, and feature usage.

The data arrives frequently and is usually retrieved by user ID or session ID. The application team needs low-latency lookups such as:

- Get all events for a user
- Get all events for a session
- Find the most recent activity for a user
- Retrieve events in timestamp order

## Sample Data Format

JSON records.

Example:

```json
{
  "event_id": "EVT00001",
  "user_id": "USR001",
  "session_id": "SES001",
  "event_type": "page_view",
  "event_timestamp": "2025-01-15T09:01:20",
  "page_name": "dashboard"
}
```

# Data Product D: Support Knowledge Base

## Scenario

The Customer Support team maintains help articles, troubleshooting guides, release notes, onboarding documentation, and internal support procedures.

Users do not usually search this data by exact ID. Instead, they ask natural-language questions such as:

- How do I reset multi-factor authentication?
- What should I do if data sync fails?
- Which plans include audit logging?
- How do I troubleshoot slow dashboard loading?

The system needs to retrieve semantically similar documents and provide relevant context to a generative AI assistant.

## Sample Data Format

Unstructured text files.

Examples:

- `reset_mfa.txt`
- `data_sync_failures.txt`
- `audit_logging_plans.txt`
- `slow_dashboard_troubleshooting.txt`

## Sample Fields or Metadata

- document_id
- document_title
- document_type
- product_area
- last_updated
- document_text
