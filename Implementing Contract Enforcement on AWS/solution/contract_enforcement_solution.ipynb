### Connect to AWS

````python
import boto3
import json
import psycopg2
import pandas as pd
from psycopg2.extras import execute_values

AWS_ACCESS_KEY_ID = "<get key from classroom>"
AWS_SECRET_ACCESS_KEY = "<get key from classroom>"
AWS_SESSION_TOKEN = "<get key from classroom>"
AWS_REGION = "us-east-1"

session = boto3.Session(
    aws_access_key_id=AWS_ACCESS_KEY_ID,
    aws_secret_access_key=AWS_SECRET_ACCESS_KEY,
    aws_session_token=AWS_SESSION_TOKEN,
    region_name=AWS_REGION,
)


# Get DB connection info from CloudFormation
cf = session.client("cloudformation")
stack = cf.describe_stacks()["Stacks"][0]
outputs = {o["OutputKey"]: o["OutputValue"] for o in stack["Outputs"]}

db_host = outputs["DBEndpoint"]
db_port = int(outputs["DBPort"])
db_name = outputs["DBName"]
secret_arn = outputs["MasterSecretArn"]

# Get credentials from Secrets Manager
sm = session.client("secretsmanager", region_name="us-east-1")
secret_response = sm.get_secret_value(SecretId=secret_arn)
db_secret = json.loads(secret_response["SecretString"])

# Connect to DB
try:
    conn = psycopg2.connect(
        host=db_host,
        port=db_port,
        dbname=db_name,
        user=db_secret["username"],
        password=db_secret["password"],
        sslmode="require",
    )
    print("Successfully connected to the database")
 
except Exception as e:
    print(f"Connection failed: {type(e).__name__}: {e}")
````

### Create Tables

````python
# Create normalized tables for the Big Box Store database
conn.rollback()
cur = conn.cursor()
create_tables_sql = """
DROP TABLE IF EXISTS order_items CASCADE;
DROP TABLE IF EXISTS orders CASCADE;
DROP TABLE IF EXISTS inventory CASCADE;
DROP TABLE IF EXISTS payment_types CASCADE;
DROP TABLE IF EXISTS distribution_center CASCADE;
DROP TABLE IF EXISTS customers CASCADE;
CREATE TABLE customers (
    customer_id VARCHAR(20) PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) NOT NULL UNIQUE,
    phone VARCHAR(25),
    loyalty_member_flag BOOLEAN NOT NULL
);
CREATE TABLE distribution_center (
    distribution_center_id VARCHAR(20) PRIMARY KEY,
    distribution_center_name VARCHAR(100) NOT NULL,
    city VARCHAR(75) NOT NULL,
    state CHAR(2) NOT NULL,
    region VARCHAR(20) NOT NULL
);
CREATE TABLE payment_types (
    payment_type_id VARCHAR(20) PRIMARY KEY,
    payment_type_name VARCHAR(50) NOT NULL,
    payment_provider VARCHAR(75)
);
CREATE TABLE inventory (
    inventory_id VARCHAR(20) PRIMARY KEY,
    product_name VARCHAR(150) NOT NULL,
    product_category VARCHAR(50) NOT NULL,
    unit_price NUMERIC(10,2) NOT NULL,
    quantity_on_hand INTEGER NOT NULL,
    distribution_center_id VARCHAR(20) NOT NULL,
    CONSTRAINT fk_inventory_distribution_center
        FOREIGN KEY (distribution_center_id)
        REFERENCES distribution_center(distribution_center_id),
    CONSTRAINT chk_inventory_price
        CHECK (unit_price > 0),
    CONSTRAINT chk_inventory_quantity
        CHECK (quantity_on_hand >= 0)
);
CREATE TABLE orders (
    order_id VARCHAR(20) PRIMARY KEY,
    order_date DATE NOT NULL,
    customer_id VARCHAR(20) NOT NULL,
    payment_type_id VARCHAR(20) NOT NULL,
    order_channel VARCHAR(50) NOT NULL,
    order_status VARCHAR(50) NOT NULL,
    order_total NUMERIC(12,2) NOT NULL,
    CONSTRAINT fk_orders_customer
        FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id),
    CONSTRAINT fk_orders_payment_type
        FOREIGN KEY (payment_type_id)
        REFERENCES payment_types(payment_type_id),
    CONSTRAINT chk_order_total
        CHECK (order_total >= 0)
);
CREATE TABLE order_items (
    order_item_id VARCHAR(20) PRIMARY KEY,
    order_id VARCHAR(20) NOT NULL,
    inventory_id VARCHAR(20) NOT NULL,
    quantity_ordered INTEGER NOT NULL,
    unit_price_at_purchase NUMERIC(10,2) NOT NULL,
    line_total NUMERIC(12,2) NOT NULL,
    CONSTRAINT fk_order_items_order
        FOREIGN KEY (order_id)
        REFERENCES orders(order_id),
    CONSTRAINT fk_order_items_inventory
        FOREIGN KEY (inventory_id)
        REFERENCES inventory(inventory_id),
    CONSTRAINT chk_quantity_ordered
        CHECK (quantity_ordered > 0),
    CONSTRAINT chk_purchase_price
        CHECK (unit_price_at_purchase > 0),
    CONSTRAINT chk_line_total
        CHECK (line_total > 0)
);
"""
cur.execute(create_tables_sql)
conn.commit()
cur.close()
print("Big Box Store tables created successfully.")
````

### Insert Data into tables

````python
conn.rollback()
cur = conn.cursor()

tables = [
    (
        "customers",
        ["customer_id", "first_name", "last_name", "email", "phone", "loyalty_member_flag"],
        customers_df,
        lambda row: (row.customer_id, row.first_name, row.last_name, row.email, row.phone, bool(row.loyalty_member_flag))
    ),
    (
        "distribution_center",
        ["distribution_center_id", "distribution_center_name", "city", "state", "region"],
        distribution_center_df,
        lambda row: (row.distribution_center_id, row.distribution_center_name, row.city, row.state, row.region)
    ),
    (
        "payment_types",
        ["payment_type_id", "payment_type_name", "payment_provider"],
        payment_types_df,
        lambda row: (row.payment_type_id, row.payment_type_name, row.payment_provider)
    ),
    (
        "inventory",
        ["inventory_id", "product_name", "product_category", "unit_price", "quantity_on_hand", "distribution_center_id"],
        inventory_df,
        lambda row: (row.inventory_id, row.product_name, row.product_category, float(row.unit_price), int(row.quantity_on_hand), row.distribution_center_id)
    ),
    (
        "orders",
        ["order_id", "order_date", "customer_id", "payment_type_id", "order_channel", "order_status", "order_total"],
        orders_df,
        lambda row: (row.order_id, row.order_date, row.customer_id, row.payment_type_id, row.order_channel, row.order_status, float(row.order_total))
    ),
    (
        "order_items",
        ["order_item_id", "order_id", "inventory_id", "quantity_ordered", "unit_price_at_purchase", "line_total"],
        order_items_df,
        lambda row: (row.order_item_id, row.order_id, row.inventory_id, int(row.quantity_ordered), float(row.unit_price_at_purchase), float(row.line_total))
    ),
]

for table, cols, df, row_fn in tables:
    sql = f"INSERT INTO {table} ({', '.join(cols)}) VALUES %s;"
    values = [row_fn(row) for row in df.itertuples(index=False)]
    execute_values(cur, sql, values)

conn.commit()
cur.close()
print("Data loaded.")
````


### Create a Trigger

Create a Trigger that updates inventory quantity once an order has been inserted into the database

---

````python
conn.rollback()
cur = conn.cursor()
cur.execute("""
CREATE OR REPLACE FUNCTION reduce_inventory_on_order_item_insert()
RETURNS TRIGGER AS $$
BEGIN
    IF (
        SELECT quantity_on_hand
        FROM inventory
        WHERE inventory_id = NEW.inventory_id
    ) < NEW.quantity_ordered THEN
        RAISE EXCEPTION 'Not enough inventory available for item %', NEW.inventory_id;
    END IF;
    UPDATE inventory
    SET quantity_on_hand = quantity_on_hand - NEW.quantity_ordered
    WHERE inventory_id = NEW.inventory_id;
    RETURN NEW;
END;
$$ LANGUAGE plpgsql;
""")
cur.execute("""
DROP TRIGGER IF EXISTS trg_reduce_inventory_on_insert
ON order_items;
CREATE TRIGGER trg_reduce_inventory_on_insert
AFTER INSERT ON order_items
FOR EACH ROW
EXECUTE FUNCTION reduce_inventory_on_order_item_insert();
""")
conn.commit()
cur.close()
print("Inventory reduction trigger created.")

# Test 1 — valid insert reduces inventory
cur = conn.cursor()
try:
    # grab a real inventory item with stock
    cur.execute("SELECT inventory_id, quantity_on_hand FROM inventory WHERE quantity_on_hand > 5 LIMIT 1;")
    inventory_id, qty_before = cur.fetchone()

    # grab real foreign keys
    cur.execute("SELECT order_id FROM orders LIMIT 1;")
    order_id = cur.fetchone()[0]

    cur.execute("""
        INSERT INTO order_items (order_item_id, order_id, inventory_id, quantity_ordered, unit_price_at_purchase, line_total)
        VALUES ('TEST001', %s, %s, 2, 9.99, 19.98);
    """, (order_id, inventory_id))

    cur.execute("SELECT quantity_on_hand FROM inventory WHERE inventory_id = %s;", (inventory_id,))
    qty_after = cur.fetchone()[0]

    conn.rollback()  # don't keep test row
    print(f"Test 1 — inventory reduction: {'PASS' if qty_after == qty_before - 2 else 'FAIL'} (before: {qty_before}, after: {qty_after})")

except Exception as e:
    conn.rollback()
    print(f"Test 1 — FAIL: {e}")
finally:
    cur.close()

# Test 2 — insert with more than available quantity should raise exception
cur = conn.cursor()
try:
    cur.execute("SELECT inventory_id, quantity_on_hand FROM inventory WHERE quantity_on_hand > 0 LIMIT 1;")
    inventory_id, qty_on_hand = cur.fetchone()

    cur.execute("SELECT order_id FROM orders LIMIT 1;")
    order_id = cur.fetchone()[0]

    cur.execute("""
        INSERT INTO order_items (order_item_id, order_id, inventory_id, quantity_ordered, unit_price_at_purchase, line_total)
        VALUES ('TEST002', %s, %s, %s, 9.99, 9.99);
    """, (order_id, inventory_id, qty_on_hand + 9999))

    conn.rollback()
    print("Test 2 — insufficient inventory check: FAIL — exception was not raised")

except Exception as e:
    conn.rollback()
    print(f"Test 2 — insufficient inventory check: PASS — correctly rejected ({e})")
finally:
    cur.close()
````


### Create a Stored Procedure

This procedure should create new ORDER and ORDER ITEM through one controlled process.

---

````python
conn.rollback()
cur = conn.cursor()
cur.execute("""
CREATE OR REPLACE PROCEDURE create_order_with_item(
    p_order_id VARCHAR,
    p_order_item_id VARCHAR,
    p_customer_id VARCHAR,
    p_payment_type_id VARCHAR,
    p_inventory_id VARCHAR,
    p_quantity_ordered INTEGER
)
LANGUAGE plpgsql
AS $$
DECLARE
    v_unit_price NUMERIC(10,2);
    v_line_total NUMERIC(12,2);
BEGIN
    SELECT unit_price
    INTO v_unit_price
    FROM inventory
    WHERE inventory_id = p_inventory_id;
    IF v_unit_price IS NULL THEN
        RAISE EXCEPTION 'Inventory item % does not exist', p_inventory_id;
    END IF;
    v_line_total := v_unit_price * p_quantity_ordered;
    INSERT INTO orders (
        order_id,
        customer_id,
        payment_type_id,
        order_total
    )
    VALUES (
        p_order_id,
        p_customer_id,
        p_payment_type_id,
        v_line_total
    );
    INSERT INTO order_items (
        order_item_id,
        order_id,
        inventory_id,
        quantity_ordered,
        unit_price_at_purchase,
        line_total
    )
    VALUES (
        p_order_item_id,
        p_order_id,
        p_inventory_id,
        p_quantity_ordered,
        v_unit_price,
        v_line_total
    );
END;
$$;
""")
conn.commit()
cur.close()
print("Stored procedure created.")

# Test 1 — valid call creates order and order_item correctly
cur = conn.cursor()
try:
    cur.execute("SELECT customer_id FROM customers LIMIT 1;")
    customer_id = cur.fetchone()[0]

    cur.execute("SELECT payment_type_id FROM payment_types LIMIT 1;")
    payment_type_id = cur.fetchone()[0]

    cur.execute("SELECT inventory_id, unit_price, quantity_on_hand FROM inventory WHERE quantity_on_hand > 2 LIMIT 1;")
    inventory_id, unit_price, qty_before = cur.fetchone()

    cur.execute("CALL create_order_with_item('PROC_TEST_01', 'PROC_ITEM_01', %s, %s, %s, 2);",
                (customer_id, payment_type_id, inventory_id))

    # verify order was created
    cur.execute("SELECT order_total FROM orders WHERE order_id = 'PROC_TEST_01';")
    order_total = cur.fetchone()[0]

    # verify order_item was created
    cur.execute("SELECT quantity_ordered, line_total FROM order_items WHERE order_item_id = 'PROC_ITEM_01';")
    quantity_ordered, line_total = cur.fetchone()

    # verify inventory was reduced
    cur.execute("SELECT quantity_on_hand FROM inventory WHERE inventory_id = %s;", (inventory_id,))
    qty_after = cur.fetchone()[0]

    expected_total = unit_price * 2
    print(f"Test 1 — order created        : {'PASS' if order_total == expected_total else 'FAIL'} (order_total: {order_total}, expected: {expected_total})")
    print(f"Test 1 — order item created   : {'PASS' if quantity_ordered == 2 and line_total == expected_total else 'FAIL'} (qty: {quantity_ordered}, line_total: {line_total})")
    print(f"Test 1 — inventory reduced    : {'PASS' if qty_after == qty_before - 2 else 'FAIL'} (before: {qty_before}, after: {qty_after})")

    conn.rollback()

except Exception as e:
    conn.rollback()
    print(f"Test 1 — FAIL: {e}")
finally:
    cur.close()

# Test 2 — invalid inventory_id should raise exception
cur = conn.cursor()
try:
    cur.execute("SELECT customer_id FROM customers LIMIT 1;")
    customer_id = cur.fetchone()[0]

    cur.execute("SELECT payment_type_id FROM payment_types LIMIT 1;")
    payment_type_id = cur.fetchone()[0]

    cur.execute("CALL create_order_with_item('PROC_TEST_02', 'PROC_ITEM_02', %s, %s, 'INVALID_ID', 1);",
                (customer_id, payment_type_id))

    conn.rollback()
    print("Test 2 — invalid inventory_id : FAIL — exception was not raised")

except Exception as e:
    conn.rollback()
    print(f"Test 2 — invalid inventory_id : PASS — correctly rejected ({e})")
finally:
    cur.close()
````
