## Create Tables

````python
# Create normalized tables for the Big Box Store database
conn = psycopg2.connect(
        host=db_host,
        port=db_port,
        dbname=db_name,
        user=db_user,
        password=db_password,
        sslmode="require",  # important for RDS
    )
cur = conn.cursor()

create_tables_sql = """
DROP TABLE IF EXISTS order_items;
DROP TABLE IF EXISTS orders;
DROP TABLE IF EXISTS inventory;
DROP TABLE IF EXISTS payment_types;
DROP TABLE IF EXISTS distribution_center;
DROP TABLE IF EXISTS customers;

CREATE TABLE customers (
    customer_id VARCHAR(20) PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) NOT NULL UNIQUE,
    phone VARCHAR(25),
    loyalty_member_flag BOOLEAN NOT NULL
);

CREATE TABLE distribution_center (
    distribution_center_id VARCHAR(20) PRIMARY KEY,
    distribution_center_name VARCHAR(100) NOT NULL,
    city VARCHAR(75) NOT NULL,
    state CHAR(2) NOT NULL,
    region VARCHAR(20) NOT NULL
);

CREATE TABLE payment_types (
    payment_type_id VARCHAR(20) PRIMARY KEY,
    payment_type_name VARCHAR(50) NOT NULL,
    payment_provider VARCHAR(75)
);

CREATE TABLE inventory (
    inventory_id VARCHAR(20) PRIMARY KEY,
    product_name VARCHAR(150) NOT NULL,
    product_category VARCHAR(50) NOT NULL,
    unit_price NUMERIC(10,2) NOT NULL,
    quantity_on_hand INTEGER NOT NULL,
    distribution_center_id VARCHAR(20) NOT NULL,

    CONSTRAINT fk_inventory_distribution_center
        FOREIGN KEY (distribution_center_id)
        REFERENCES distribution_center(distribution_center_id),

    CONSTRAINT chk_inventory_price
        CHECK (unit_price > 0),

    CONSTRAINT chk_inventory_quantity
        CHECK (quantity_on_hand >= 0)
);

CREATE TABLE orders (
    order_id VARCHAR(20) PRIMARY KEY,
    order_date DATE NOT NULL,
    customer_id VARCHAR(20) NOT NULL,
    payment_type_id VARCHAR(20) NOT NULL,
    order_channel VARCHAR(50) NOT NULL,
    order_status VARCHAR(50) NOT NULL,
    order_total NUMERIC(12,2) NOT NULL,

    CONSTRAINT fk_orders_customer
        FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id),

    CONSTRAINT fk_orders_payment_type
        FOREIGN KEY (payment_type_id)
        REFERENCES payment_types(payment_type_id),

    CONSTRAINT chk_order_total
        CHECK (order_total >= 0)
);

CREATE TABLE order_items (
    order_item_id VARCHAR(20) PRIMARY KEY,
    order_id VARCHAR(20) NOT NULL,
    inventory_id VARCHAR(20) NOT NULL,
    quantity_ordered INTEGER NOT NULL,
    unit_price_at_purchase NUMERIC(10,2) NOT NULL,
    line_total NUMERIC(12,2) NOT NULL,

    CONSTRAINT fk_order_items_order
        FOREIGN KEY (order_id)
        REFERENCES orders(order_id),

    CONSTRAINT fk_order_items_inventory
        FOREIGN KEY (inventory_id)
        REFERENCES inventory(inventory_id),

    CONSTRAINT chk_quantity_ordered
        CHECK (quantity_ordered > 0),

    CONSTRAINT chk_purchase_price
        CHECK (unit_price_at_purchase > 0),

    CONSTRAINT chk_line_total
        CHECK (line_total > 0)
);
"""

cur.execute(create_tables_sql)
conn.commit()
cur.close()

print("Big Box Store tables created successfully.")
````

## Load CSV files into memory

````python
import pandas as pd
from psycopg2.extras import execute_values

conn.rollback()

# Read CSV files
customers_df = pd.read_csv("customers.csv")
distribution_center_df = pd.read_csv("distribution_center.csv")
payment_types_df = pd.read_csv("payment_types.csv")
inventory_df = pd.read_csv("inventory.csv")
orders_df = pd.read_csv("orders.csv")
order_items_df = pd.read_csv("order_items.csv")

print("customers:", len(customers_df))
print("distribution_center:", len(distribution_center_df))
print("payment_types:", len(payment_types_df))
print("inventory:", len(inventory_df))
print("orders:", len(orders_df))
print("order_items:", len(order_items_df))
````

## Clear any existing data

````python
# Optional: clear existing data before loading
# Child tables first because of foreign keys

conn.rollback()
cur = conn.cursor()

cur.execute("""
TRUNCATE TABLE order_items CASCADE;
TRUNCATE TABLE orders CASCADE;
TRUNCATE TABLE inventory CASCADE;
TRUNCATE TABLE payment_types CASCADE;
TRUNCATE TABLE distribution_center CASCADE;
TRUNCATE TABLE customers CASCADE;
""")

conn.commit()
cur.close()

print("Existing data cleared.")
````

## Insert Data into tables

````python
# Load customers

conn.rollback()
cur = conn.cursor()

sql = """
INSERT INTO customers (
    customer_id,
    first_name,
    last_name,
    email,
    phone,
    loyalty_member_flag
)
VALUES %s;
"""

values = [
    (
        row.customer_id,
        row.first_name,
        row.last_name,
        row.email,
        row.phone,
        bool(row.loyalty_member_flag)
    )
    for row in customers_df.itertuples(index=False)
]

execute_values(cur, sql, values)
conn.commit()
cur.close()

print("customers loaded.")
````
