## Connect to AWS



````python
import boto3
import json
import psycopg2
import pandas as pd
from psycopg2.extras import execute_values


AWS_ACCESS_KEY_ID = "<get key from classroom>"
AWS_SECRET_ACCESS_KEY = "<get key from classroom>"
AWS_SESSION_TOKEN = "<get key from classroom>"
AWS_REGION = "us-east-1"

session = boto3.Session(
    aws_access_key_id=AWS_ACCESS_KEY_ID,
    aws_secret_access_key=AWS_SECRET_ACCESS_KEY,
    aws_session_token=AWS_SESSION_TOKEN,
    region_name=AWS_REGION,
)


# Get DB connection info from CloudFormation
cf = session.client("cloudformation")
stack = cf.describe_stacks()["Stacks"][0]
outputs = {o["OutputKey"]: o["OutputValue"] for o in stack["Outputs"]}

db_host = outputs["DBEndpoint"]
db_port = int(outputs["DBPort"])
db_name = outputs["DBName"]
secret_arn = outputs["MasterSecretArn"]

# Get credentials from Secrets Manager
sm = session.client("secretsmanager", region_name="us-east-1")
secret_response = sm.get_secret_value(SecretId=secret_arn)
db_secret = json.loads(secret_response["SecretString"])

# Connect to DB
try:
    conn = psycopg2.connect(
        host=db_host,
        port=db_port,
        dbname=db_name,
        user=db_secret["username"],
        password=db_secret["password"],
        sslmode="require",
    )
    print("Successfully connected to the database")
 
except Exception as e:
    print(f"Connection failed: {type(e).__name__}: {e}")
````

### Create Tables

````python
# Create normalized tables for the Big Box Store database
conn.rollback()
cur = conn.cursor()
create_tables_sql = """
DROP TABLE IF EXISTS order_items CASCADE;
DROP TABLE IF EXISTS orders CASCADE;
DROP TABLE IF EXISTS inventory CASCADE;
DROP TABLE IF EXISTS payment_types CASCADE;
DROP TABLE IF EXISTS distribution_center CASCADE;
DROP TABLE IF EXISTS customers CASCADE;
CREATE TABLE customers (
    customer_id VARCHAR(20) PRIMARY KEY,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) NOT NULL UNIQUE,
    phone VARCHAR(25),
    loyalty_member_flag BOOLEAN NOT NULL
);
CREATE TABLE distribution_center (
    distribution_center_id VARCHAR(20) PRIMARY KEY,
    distribution_center_name VARCHAR(100) NOT NULL,
    city VARCHAR(75) NOT NULL,
    state CHAR(2) NOT NULL,
    region VARCHAR(20) NOT NULL
);
CREATE TABLE payment_types (
    payment_type_id VARCHAR(20) PRIMARY KEY,
    payment_type_name VARCHAR(50) NOT NULL,
    payment_provider VARCHAR(75)
);
CREATE TABLE inventory (
    inventory_id VARCHAR(20) PRIMARY KEY,
    product_name VARCHAR(150) NOT NULL,
    product_category VARCHAR(50) NOT NULL,
    unit_price NUMERIC(10,2) NOT NULL,
    quantity_on_hand INTEGER NOT NULL,
    distribution_center_id VARCHAR(20) NOT NULL,
    CONSTRAINT fk_inventory_distribution_center
        FOREIGN KEY (distribution_center_id)
        REFERENCES distribution_center(distribution_center_id),
    CONSTRAINT chk_inventory_price
        CHECK (unit_price > 0),
    CONSTRAINT chk_inventory_quantity
        CHECK (quantity_on_hand >= 0)
);
CREATE TABLE orders (
    order_id VARCHAR(20) PRIMARY KEY,
    order_date DATE NOT NULL,
    customer_id VARCHAR(20) NOT NULL,
    payment_type_id VARCHAR(20) NOT NULL,
    order_channel VARCHAR(50) NOT NULL,
    order_status VARCHAR(50) NOT NULL,
    order_total NUMERIC(12,2) NOT NULL,
    CONSTRAINT fk_orders_customer
        FOREIGN KEY (customer_id)
        REFERENCES customers(customer_id),
    CONSTRAINT fk_orders_payment_type
        FOREIGN KEY (payment_type_id)
        REFERENCES payment_types(payment_type_id),
    CONSTRAINT chk_order_total
        CHECK (order_total >= 0)
);
CREATE TABLE order_items (
    order_item_id VARCHAR(20) PRIMARY KEY,
    order_id VARCHAR(20) NOT NULL,
    inventory_id VARCHAR(20) NOT NULL,
    quantity_ordered INTEGER NOT NULL,
    unit_price_at_purchase NUMERIC(10,2) NOT NULL,
    line_total NUMERIC(12,2) NOT NULL,
    CONSTRAINT fk_order_items_order
        FOREIGN KEY (order_id)
        REFERENCES orders(order_id),
    CONSTRAINT fk_order_items_inventory
        FOREIGN KEY (inventory_id)
        REFERENCES inventory(inventory_id),
    CONSTRAINT chk_quantity_ordered
        CHECK (quantity_ordered > 0),
    CONSTRAINT chk_purchase_price
        CHECK (unit_price_at_purchase > 0),
    CONSTRAINT chk_line_total
        CHECK (line_total > 0)
);
"""
cur.execute(create_tables_sql)
conn.commit()
cur.close()
print("Big Box Store tables created successfully.")
````

### Load CSV files into memory

````python
# Read CSV files
customers_df = pd.read_csv("../serving-layer-exercise/customers.csv")
distribution_center_df = pd.read_csv("../serving-layer-exercise/distribution_center.csv")
payment_types_df = pd.read_csv("../serving-layer-exercise/payment_types.csv")
inventory_df = pd.read_csv("../serving-layer-exercise/inventory.csv")
orders_df = pd.read_csv("../serving-layer-exercise/orders.csv")
order_items_df = pd.read_csv("../serving-layer-exercise/order_items.csv")

print("customers:", len(customers_df))
print("distribution_center:", len(distribution_center_df))
print("payment_types:", len(payment_types_df))
print("inventory:", len(inventory_df))
print("orders:", len(orders_df))
print("order_items:", len(order_items_df))
````

### Insert Data into tables

````python
conn.rollback()
cur = conn.cursor()

tables = [
    (
        "customers",
        ["customer_id", "first_name", "last_name", "email", "phone", "loyalty_member_flag"],
        customers_df,
        lambda row: (row.customer_id, row.first_name, row.last_name, row.email, row.phone, bool(row.loyalty_member_flag))
    ),
    (
        "distribution_center",
        ["distribution_center_id", "distribution_center_name", "city", "state", "region"],
        distribution_center_df,
        lambda row: (row.distribution_center_id, row.distribution_center_name, row.city, row.state, row.region)
    ),
    (
        "payment_types",
        ["payment_type_id", "payment_type_name", "payment_provider"],
        payment_types_df,
        lambda row: (row.payment_type_id, row.payment_type_name, row.payment_provider)
    ),
    (
        "inventory",
        ["inventory_id", "product_name", "product_category", "unit_price", "quantity_on_hand", "distribution_center_id"],
        inventory_df,
        lambda row: (row.inventory_id, row.product_name, row.product_category, float(row.unit_price), int(row.quantity_on_hand), row.distribution_center_id)
    ),
    (
        "orders",
        ["order_id", "order_date", "customer_id", "payment_type_id", "order_channel", "order_status", "order_total"],
        orders_df,
        lambda row: (row.order_id, row.order_date, row.customer_id, row.payment_type_id, row.order_channel, row.order_status, float(row.order_total))
    ),
    (
        "order_items",
        ["order_item_id", "order_id", "inventory_id", "quantity_ordered", "unit_price_at_purchase", "line_total"],
        order_items_df,
        lambda row: (row.order_item_id, row.order_id, row.inventory_id, int(row.quantity_ordered), float(row.unit_price_at_purchase), float(row.line_total))
    ),
]

for table, cols, df, row_fn in tables:
    sql = f"INSERT INTO {table} ({', '.join(cols)}) VALUES %s;"
    values = [row_fn(row) for row in df.itertuples(index=False)]
    execute_values(cur, sql, values)

conn.commit()
cur.close()
print("Data loaded.")
````


### Create Customer Facing View

Create a view that will display order information in a customer friendly format

This will be way the customer interacts with the database instead of viewing the underlying tables

---

````python
conn.rollback()
cur = conn.cursor()
cur.execute("""
CREATE OR REPLACE VIEW vw_customer_order_summary AS
SELECT
    o.order_id,
    o.order_date,
    c.customer_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    c.email,
    o.order_channel,
    o.order_status,
    pt.payment_type_name,
    o.order_total
FROM orders o
JOIN customers c
    ON o.customer_id = c.customer_id
JOIN payment_types pt
    ON o.payment_type_id = pt.payment_type_id;
""")
conn.commit()
cur.close()
print("Serving layer view created.")


df = pd.read_sql("SELECT * FROM vw_customer_order_summary LIMIT 10;", conn)
display(df)
````

### Alter Underlying Tables

Alter an underlying table in your view by adding a column to demostrate that your view is not affected by changes to the base table

````python
conn.rollback()
cur = conn.cursor()

cur.execute("""
ALTER TABLE orders
ADD COLUMN IF NOT EXISTS fulfillment_priority VARCHAR(20)
DEFAULT 'Standard';
""")

conn.commit()
cur.close()

print("Internal table changed.")

df = pd.read_sql("SELECT * FROM vw_customer_order_summary LIMIT 10;", conn)
display(df)
````

### Add Logic Layer to View

Modify your view to add a WHERE clause that provides added logic to your view without altering the look of output columns for the customer

---

````python
conn.rollback()
cur = conn.cursor()

cur.execute("""
CREATE OR REPLACE VIEW vw_customer_order_summary AS
SELECT
    o.order_id,
    o.order_date,
    c.customer_id,
    c.first_name || ' ' || c.last_name AS customer_name,
    c.email,
    o.order_channel,
    o.order_status,
    pt.payment_type_name,
    o.order_total
FROM orders o
JOIN customers c
    ON o.customer_id = c.customer_id
JOIN payment_types pt
    ON o.payment_type_id = pt.payment_type_id
WHERE o.order_status <> 'Cancelled';
""")

conn.commit()
cur.close()

print("Serving layer view updated internally.")

df = pd.read_sql("SELECT * FROM vw_customer_order_summary LIMIT 10;", conn)
display(df)
````
